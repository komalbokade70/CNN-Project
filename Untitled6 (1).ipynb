# Import

```python
import os, shutil, random, subprocess, zipfile
import xml.etree.ElementTree as ET

import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from PIL import Image, ImageFile

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from tensorflow.keras.applications import MobileNetV2
from tensorflow.keras.applications.mobilenet_v2 import preprocess_input
from tensorflow.keras.callbacks import EarlyStopping, ModelCheckpoint
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (confusion_matrix, classification_report,
                             roc_auc_score, average_precision_score,
                             precision_recall_curve)

ImageFile.LOAD_TRUNCATED_IMAGES = True
```

```python
SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)

BASE_DIR        = "/content"                # Colab working dir
IMAGES_DIR      = f"{BASE_DIR}/images"
ANNOTATIONS_DIR = f"{BASE_DIR}/annotations"
DATASET_DIR     = f"{BASE_DIR}/dataset"

IMG_SIZE   = (160, 160)
BATCH_SIZE = 32
MIN_FACE   = 24      # skip faces smaller than this many pixels (width or height)
PAD        = 0.15    # extra context around each face box
SPLIT      = (0.70, 0.15, 0.15)   # train / val / test, split by source image
os.makedirs("assets", exist_ok=True)
```

# Data Download

```python
from google.colab import files

if not os.path.exists(f"{BASE_DIR}/face-mask-detection.zip"):
    files.upload()   # choose kaggle.json
    os.makedirs(os.path.expanduser("~/.kaggle"), exist_ok=True)
    shutil.copy("kaggle.json", os.path.expanduser("~/.kaggle/kaggle.json"))
    os.chmod(os.path.expanduser("~/.kaggle/kaggle.json"), 0o600)
    subprocess.run(["kaggle", "datasets", "download", "-d",
                    "andrewmvd/face-mask-detection", "-p", BASE_DIR], check=True)

with zipfile.ZipFile(f"{BASE_DIR}/face-mask-detection.zip") as z:
    z.extractall(BASE_DIR)
print(os.listdir(BASE_DIR))
```

# Data Loading

```python
def build_dataset():
    if os.path.exists(DATASET_DIR):
        shutil.rmtree(DATASET_DIR)

    xml_files = sorted(f for f in os.listdir(ANNOTATIONS_DIR) if f.endswith(".xml"))
    random.Random(SEED).shuffle(xml_files)

    n = len(xml_files)
    n_train, n_val = int(SPLIT[0] * n), int(SPLIT[1] * n)
    split_of = {}
    for i, f in enumerate(xml_files):
        split_of[f] = "train" if i < n_train else "val" if i < n_train + n_val else "test"

    for s in ("train", "val", "test"):
        for c in ("with_mask", "without_mask"):
            os.makedirs(f"{DATASET_DIR}/{s}/{c}", exist_ok=True)

    counts, skipped_small, skipped_incorrect = {}, 0, 0
    for xml_file in xml_files:
        root = ET.parse(os.path.join(ANNOTATIONS_DIR, xml_file)).getroot()
        filename = root.find("filename").text
        img_path = os.path.join(IMAGES_DIR, filename)
        if not os.path.exists(img_path):
            continue
        img = Image.open(img_path).convert("RGB")
        W, H = img.size
        split = split_of[xml_file]

        for i, obj in enumerate(root.findall("object")):
            label = obj.find("name").text
            if label not in ("with_mask", "without_mask"):
                skipped_incorrect += 1          # drops mask_weared_incorrect
                continue
            bb = obj.find("bndbox")
            xmin, ymin = int(bb.find("xmin").text), int(bb.find("ymin").text)
            xmax, ymax = int(bb.find("xmax").text), int(bb.find("ymax").text)
            bw, bh = xmax - xmin, ymax - ymin
            if min(bw, bh) < MIN_FACE:
                skipped_small += 1
                continue
            px, py = int(PAD * bw), int(PAD * bh)
            box = (max(0, xmin - px), max(0, ymin - py),
                   min(W, xmax + px), min(H, ymax + py))
            out = f"{DATASET_DIR}/{split}/{label}/{os.path.splitext(filename)[0]}_{i}.jpg"
            img.crop(box).save(out, quality=95)
            counts[(split, label)] = counts.get((split, label), 0) + 1

    print("Skipped (too small):", skipped_small)
    print("Skipped (mask_weared_incorrect):", skipped_incorrect)
    for s in ("train", "val", "test"):
        print(s, {c: counts.get((s, c), 0) for c in ("with_mask", "without_mask")})

build_dataset()
```

# Data Generator

```python
train_datagen = ImageDataGenerator(
    preprocessing_function=preprocess_input,
    rotation_range=15,
    zoom_range=0.2,
    shear_range=0.1,
    brightness_range=(0.7, 1.3),
    horizontal_flip=True,
)
eval_datagen = ImageDataGenerator(preprocessing_function=preprocess_input)

train_data = train_datagen.flow_from_directory(
    f"{DATASET_DIR}/train", target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode="binary", shuffle=True, seed=SEED)
val_data = eval_datagen.flow_from_directory(
    f"{DATASET_DIR}/val", target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode="binary", shuffle=False)
test_data = eval_datagen.flow_from_directory(
    f"{DATASET_DIR}/test", target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode="binary", shuffle=False)

print("Classes:", train_data.class_indices)     # with_mask=0, without_mask=1
print("Train counts:", np.bincount(train_data.classes))
print("Val counts:  ", np.bincount(val_data.classes))
print("Test counts: ", np.bincount(test_data.classes))
print(f"Majority-class baseline on test: {np.bincount(test_data.classes).max() / len(test_data.classes):.3f}")
```

# Model




```python
base_model = MobileNetV2(input_shape=IMG_SIZE + (3,), include_top=False, weights="imagenet")
base_model.trainable = False

inputs = keras.Input(shape=IMG_SIZE + (3,))
x = base_model(inputs, training=False)      # keeps BatchNorm in inference mode, also during fine-tuning
x = layers.GlobalAveragePooling2D()(x)
x = layers.Dropout(0.3)(x)
x = layers.Dense(128, activation="relu")(x)
x = layers.Dropout(0.3)(x)
outputs = layers.Dense(1, activation="sigmoid")(x)
model = keras.Model(inputs, outputs)

METRICS = ["accuracy",
           keras.metrics.AUC(name="auc"),
           keras.metrics.Precision(name="precision"),
           keras.metrics.Recall(name="recall")]

model.compile(optimizer=keras.optimizers.Adam(1e-3),
              loss="binary_crossentropy", metrics=METRICS)
model.summary()
```

# Train the classification head

```python
classes = np.unique(train_data.classes)
weights = compute_class_weight("balanced", classes=classes, y=train_data.classes)
class_weight = dict(zip(classes, weights))
print("Class weights:", class_weight)

STAGE1_EPOCHS = 30
cb1 = [
    EarlyStopping(monitor="val_auc", mode="max", patience=5, restore_best_weights=True),
    ModelCheckpoint("stage1.keras", monitor="val_auc", mode="max", save_best_only=True),
]
history = model.fit(train_data, epochs=STAGE1_EPOCHS, validation_data=val_data,
                    class_weight=class_weight, callbacks=cb1)
stage1_epochs = len(history.history["loss"])
```

# fine-tune the last 30 layers (BatchNorm stays frozen)

```python
base_model.trainable = True
for layer in base_model.layers[:-30]:
    layer.trainable = False
for layer in base_model.layers[-30:]:
    if isinstance(layer, layers.BatchNormalization):
        layer.trainable = False             # avoids the loss spike from unfreezing BN

model.compile(optimizer=keras.optimizers.Adam(1e-5),
              loss="binary_crossentropy", metrics=METRICS)

FT_EPOCHS = 15
cb2 = [
    EarlyStopping(monitor="val_auc", mode="max", patience=4, restore_best_weights=True),
    ModelCheckpoint("stage2.keras", monitor="val_auc", mode="max", save_best_only=True),
]
history_fine = model.fit(train_data, epochs=stage1_epochs + FT_EPOCHS,
                         initial_epoch=stage1_epochs, validation_data=val_data,
                         class_weight=class_weight, callbacks=cb2)
```

# Keep whichever stage is better on validation

```python
m1 = keras.models.load_model("stage1.keras")
m2 = keras.models.load_model("stage2.keras")
r1 = m1.evaluate(val_data, return_dict=True, verbose=0)
r2 = m2.evaluate(val_data, return_dict=True, verbose=0)
print("Stage 1 val:", {k: round(v, 4) for k, v in r1.items()})
print("Stage 2 val:", {k: round(v, 4) for k, v in r2.items()})

model = m2 if r2["auc"] > r1["auc"] else m1
print("Using:", "stage 2 (fine-tuned)" if model is m2 else "stage 1 (frozen base)")
```

# Fair train vs val vs test comparison

```python
import json
train_eval = eval_datagen.flow_from_directory(
    f"{DATASET_DIR}/train", target_size=IMG_SIZE, batch_size=BATCH_SIZE,
    class_mode="binary", shuffle=False)

results = {}
for name, gen in [("train", train_eval), ("val", val_data), ("test", test_data)]:
    r = model.evaluate(gen, return_dict=True, verbose=0)
    results[name] = {k: round(float(v), 4) for k, v in r.items()}
    print(f"{name:5s}", results[name])

gap = results["train"]["accuracy"] - results["val"]["accuracy"]
print(f"\nTrain - Val accuracy gap: {gap:+.3f}")
print("Positive gap = mild overfitting; near zero = good fit; large positive (> 0.05) = overfitting.")
```

# Training curve

```python
def joined(key):
    return history.history[key] + history_fine.history[key]

fig, ax = plt.subplots(1, 3, figsize=(16, 4))
for a, (k, title) in zip(ax, [("accuracy", "Accuracy"), ("loss", "Loss"), ("auc", "AUC")]):
    a.plot(joined(k), label=f"train {k}")
    a.plot(joined(f"val_{k}"), label=f"val {k}")
    a.axvline(stage1_epochs - 1, color="gray", linestyle="--", label="fine-tune start")
    a.set_title(f"{title} over epochs"); a.set_xlabel("epoch"); a.legend()
plt.tight_layout()
plt.savefig("assets/training_curves.png", dpi=150, bbox_inches="tight")
plt.show()
```

#Tune the decision threshold on validation

```python
val_probs = model.predict(val_data, verbose=0).ravel()
prec, rec, thr = precision_recall_curve(val_data.classes, val_probs)
f1 = 2 * prec[:-1] * rec[:-1] / (prec[:-1] + rec[:-1] + 1e-9)
best_thr = float(thr[np.argmax(f1)])
print(f"Best threshold on val (max F1 for without_mask): {best_thr:.3f}")
```

# Final evaluation on the held-out test set

```python
test_probs = model.predict(test_data, verbose=0).ravel()
y_true = test_data.classes
names = list(test_data.class_indices.keys())

for thr_value, tag in [(0.5, "default threshold 0.5"), (best_thr, f"tuned threshold {best_thr:.2f}")]:
    y_pred = (test_probs > thr_value).astype(int)
    print(f"===== {tag} =====")
    print(classification_report(y_true, y_pred, target_names=names, digits=3))

print(f"ROC-AUC: {roc_auc_score(y_true, test_probs):.3f}")
print(f"PR-AUC (without_mask): {average_precision_score(y_true, test_probs):.3f}")

y_pred = (test_probs > best_thr).astype(int)
results["test_tuned_threshold"] = {"threshold": round(best_thr, 3),
    "roc_auc": round(float(roc_auc_score(y_true, test_probs)), 4),
    "pr_auc": round(float(average_precision_score(y_true, test_probs)), 4)}
json.dump(results, open("assets/results.json", "w"), indent=2)
cm = confusion_matrix(y_true, y_pred)
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=names, yticklabels=names)
plt.xlabel("Predicted"); plt.ylabel("True"); plt.title("Confusion matrix (test)")
plt.savefig("assets/confusion_matrix.png", dpi=150, bbox_inches="tight")
plt.show()
```

# Sample predictions (random test images)

```python
idx = np.random.choice(len(test_data.filepaths), 12, replace=False)
plt.figure(figsize=(12, 9))
for k, i in enumerate(idx):
    img = Image.open(test_data.filepaths[i]).convert("RGB")
    p = test_probs[i]
    pred = names[int(p > best_thr)]
    true = names[y_true[i]]
    plt.subplot(3, 4, k + 1)
    plt.imshow(img); plt.axis("off")
    plt.title(f"true: {true}\npred: {pred} ({p:.2f})",
              color="green" if pred == true else "red", fontsize=9)
plt.tight_layout()
plt.savefig("assets/sample_predictions.png", dpi=150, bbox_inches="tight")
plt.show()
```

# Save the model and predict on a new image

```python
model.save("mask_detector.keras")

def predict_image(path, threshold=best_thr):
    img = Image.open(path).convert("RGB").resize(IMG_SIZE)
    x = preprocess_input(np.array(img, dtype="float32"))[None, ...]
    p = float(model.predict(x, verbose=0)[0][0])
    return ("without_mask" if p > threshold else "with_mask"), p

# print(predict_image("/content/some_face.jpg"))
```
